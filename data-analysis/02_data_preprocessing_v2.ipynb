```python
def check_multi_values(df, video_col="file_nm"):
    result = []

    for col in df.columns:
        if col == video_col:
            continue

        max_unique = df.groupby(video_col)[col].nunique().max()

        if max_unique > 1:
            result.append({
                "column": col,
                "max_values_per_video": max_unique
            })

    return pd.DataFrame(result).sort_values(
        "max_values_per_video",
        ascending=False
    )


print("[API 1]")
display(check_multi_values(api1))

print("[API 3]")
display(check_multi_values(api3))
```

```python
# 최종 필터링된 데이터에서 같은 영상(file_nm) 내 여러 값을 가지는 컬럼 확인

def check_multi_values(df, video_col="file_nm"):
    result = []

    for col in df.columns:
        if col == video_col:
            continue

        max_unique = df.groupby(video_col)[col].nunique().max()

        if max_unique > 1:
            result.append({
                "column": col,
                "max_values_per_video": max_unique
            })

    return pd.DataFrame(result).sort_values(
        "max_values_per_video",
        ascending=False
    )

print("[API 1]")
display(check_multi_values(api1_filtered))

print("[API 3]")
display(check_multi_values(api3_filtered))
```

```python
# file_nm 기준으로 여러 행을 영상당 1행으로 집계하고 실제 영상 URL 생성

multi_cols_api1 = [
    "trng_mscl_nm",
    "rptt_tcnt_nm",
    "ftns_lvl_nm",
    "tool_nm",
    "ecrg_cycl_nm",
    "trng_mscl_zn_nm",
    "trng_plc_nm",
    "trng_mscl_part",
    "trng_nm",
    "ftns_fctr_nm",
    "trng_mscl_class",
    "trng_hr_nm",
    "trng_mscl_eng_nm",
    "set_cnt_nm"
]

multi_cols_api3 = [
    "trng_mscl_nm",
    "rptt_tcnt_nm",
    "tool_nm",
    "ecrg_cycl_nm",
    "trng_mscl_zn_nm",
    "trng_plc_nm",
    "trng_mscl_part",
    "trng_nm",
    "trng_mscl_class",
    "trng_hr_nm",
    "trng_mscl_eng_nm",
    "set_cnt_nm"
]

def join_unique(series):
    values = series.dropna().astype(str).unique()
    return "|".join(values)

def aggregate_videos(df, multi_cols):
    agg_dict = {}

    for col in df.columns:
        if col == "file_nm":
            continue

        if col in multi_cols:
            agg_dict[col] = join_unique
        else:
            agg_dict[col] = "first"

    result = (
        df.groupby("file_nm", as_index=False)
        .agg(agg_dict)
    )

    result["video_url"] = (
        result["file_url"].fillna("")
        + result["file_nm"].fillna("")
    )

    return result


api1_video = aggregate_videos(api1_filtered, multi_cols_api1)
api3_video = aggregate_videos(api3_filtered, multi_cols_api3)

print("API 1 영상:", len(api1_video))
print("API 3 영상:", len(api3_video))
print("총 영상:", len(api1_video) + len(api3_video))
```

```python
# 제외 도구명이 운동명 또는 영상 제목에 포함됐지만 tool_nm에는 제대로 기록되지 않은 후보 영상을 확인

def find_excluded_tool_in_text(df):
    results = []

    for _, row in df.iterrows():
        text = " ".join([
            str(row.get("vdo_ttl_nm", "")),
            str(row.get("trng_nm", ""))
        ])

        matched_tools = [
            tool for tool in excluded_tools
            if tool in text
        ]

        if matched_tools:
            results.append({
                "file_nm": row["file_nm"],
                "vdo_ttl_nm": row["vdo_ttl_nm"],
                "trng_nm": row["trng_nm"],
                "tool_nm": row["tool_nm"],
                "matched_excluded_tools": "|".join(matched_tools)
            })

    return pd.DataFrame(results)


api1_text_check = find_excluded_tool_in_text(api1_video)
api3_text_check = find_excluded_tool_in_text(api3_video)

print("[API 1]")
display(api1_text_check)

print("\n[API 3]")
display(api3_text_check)
```

```python
# 최종 추천 후보 영상의 길이를 1분 단위 구간으로 나누어 API별 영상 수와 비율을 표로 확인

def video_length_table(df):
    bins = [0, 60, 120, 180, 240, 300, 360, 420, 480, 540, 600]
    labels = [
        "1분 미만",
        "1~2분",
        "2~3분",
        "3~4분",
        "4~5분",
        "5~6분",
        "6~7분",
        "7~8분",
        "8~9분",
        "9~10분"
    ]

    length_group = pd.cut(
        df["vdo_len"],
        bins=bins,
        labels=labels,
        right=False
    )

    result = (
        length_group
        .value_counts(sort=False)
        .rename("영상 수")
        .to_frame()
    )

    result["비율(%)"] = (
        result["영상 수"] / len(df) * 100
    ).round(1)

    return result


print("[API 1]")
display(video_length_table(api1_video))

print("[API 3]")
display(video_length_table(api3_video))
```

```python
# API 1의 체력요인명을 기존 최종 CSV의 6개 체력요인 가중치로 변환하기 위한 매핑 정의

factor_mapping = {
    "근력": {
        "strength": 1.0
    },
    "근력/근지구력": {
        "strength": 0.5,
        "muscularEndurance": 0.5
    },
    "심폐지구력": {
        "cardiovascularEndurance": 1.0
    },
    "전신지구력": {
        "cardiovascularEndurance": 1.0
    },
    "유산소": {
        "cardiovascularEndurance": 1.0
    },
    "유연성": {
        "flexibility": 1.0
    },
    "민첩성": {
        "agility": 1.0
    },
    "민첩성/순발력": {
        "agility": 0.5,
        "power": 0.5
    },
    "순발력": {
        "power": 1.0
    }
}

factor_cols = [
    "strength",
    "muscularEndurance",
    "cardiovascularEndurance",
    "flexibility",
    "agility",
    "power"
]

print("체력요인 매핑:")
for factor, weights in factor_mapping.items():
    print(f"{factor}: {weights}")
```

```python
# 현재 API 1 후보 283개의 체력요인 분포와 6개 체력요인 매핑에서 처리되지 않는 값을 확인

api1_factor_check = (
    api1_video["ftns_fctr_nm"]
    .fillna("")
    .str.split("|")
    .explode()
    .str.strip()
)

api1_factor_check = api1_factor_check[
    api1_factor_check != ""
]

factor_distribution = (
    api1_factor_check
    .value_counts()
    .rename_axis("ftns_fctr_nm")
    .reset_index(name="영상 등장 횟수")
)

factor_distribution["mapping_status"] = (
    factor_distribution["ftns_fctr_nm"]
    .apply(
        lambda x: "매핑 가능"
        if x in factor_mapping
        else "매핑 필요"
    )
)

display(factor_distribution)

unmapped_factors = sorted(
    set(api1_factor_check.unique())
    - set(factor_mapping.keys())
)

print("매핑되지 않은 체력요인:", unmapped_factors)
```

```python
# API 1의 체력요인을 6개 체력요인 가중치로 변환하고 영상별 합이 1이 되도록 정규화

def calculate_factor_weights(factor_text):
    weights = {col: 0.0 for col in factor_cols}

    if pd.isna(factor_text) or factor_text == "":
        return pd.Series(weights)

    factors = [
        factor.strip()
        for factor in str(factor_text).split("|")
        if factor.strip()
    ]

    # 매핑되지 않은 체력요인이 하나라도 있으면 일단 계산하지 않음
    if any(factor not in factor_mapping for factor in factors):
        return pd.Series({col: None for col in factor_cols})

    for factor in factors:
        for col, value in factor_mapping[factor].items():
            weights[col] += value

    total = sum(weights.values())

    if total > 0:
        weights = {
            col: value / total
            for col, value in weights.items()
        }

    return pd.Series(weights)


api1_factor_weights = (
    api1_video["ftns_fctr_nm"]
    .apply(calculate_factor_weights)
)

api1_factor_result = pd.concat(
    [
        api1_video[["file_nm", "vdo_ttl_nm", "ftns_fctr_nm"]].reset_index(drop=True),
        api1_factor_weights.reset_index(drop=True)
    ],
    axis=1
)

print("API 1 전체 영상:", len(api1_factor_result))

print(
    "6요인 계산 완료:",
    api1_factor_result[factor_cols]
    .notna()
    .all(axis=1)
    .sum()
)

print(
    "추가 처리 필요:",
    api1_factor_result[factor_cols]
    .isna()
    .any(axis=1)
    .sum()
)

display(
    api1_factor_result[
        api1_factor_result[factor_cols]
        .isna()
        .any(axis=1)
    ][
        ["file_nm", "vdo_ttl_nm", "ftns_fctr_nm"]
    ]
)
```

```python
# 현재 API 3 후보 112개의 운동명을 API 1의 운동명-체력요인 정보와 매칭하여 영상별 매칭 가능 여부를 확인

api3_current = api3[
    api3["file_nm"].isin(api3_video["file_nm"])
].copy()

api1_exercise_factor = (
    api1[
        ["trng_nm", "ftns_fctr_nm"]
    ]
    .dropna(subset=["trng_nm", "ftns_fctr_nm"])
    .drop_duplicates()
    .copy()
)

# 동일 운동명이 여러 체력요인으로 분류된 경우를 확인
api1_exercise_factor_count = (
    api1_exercise_factor
    .groupby("trng_nm")["ftns_fctr_nm"]
    .nunique()
)

# 체력요인이 하나로 일관된 운동명만 매칭 기준으로 사용
stable_exercise_names = (
    api1_exercise_factor_count[
        api1_exercise_factor_count == 1
    ]
    .index
)

api1_stable_factor = (
    api1_exercise_factor[
        api1_exercise_factor["trng_nm"].isin(stable_exercise_names)
    ]
    .drop_duplicates(subset=["trng_nm"])
    .set_index("trng_nm")["ftns_fctr_nm"]
)

api3_current["matched_factor"] = (
    api3_current["trng_nm"]
    .map(api1_stable_factor)
)

# 영상별 구성 운동 수와 체력요인 매칭 여부를 확인
api3_match_check = (
    api3_current
    .groupby("file_nm")
    .agg(
        exercise_count=("trng_nm", "nunique"),
        matched_exercise_count=(
            "matched_factor",
            lambda x: x.notna().groupby(
                api3_current.loc[x.index, "trng_nm"]
            ).max().sum()
        )
    )
    .reset_index()
)

api3_match_check["all_matched"] = (
    api3_match_check["exercise_count"]
    == api3_match_check["matched_exercise_count"]
)

print("API 3 후보 영상:", len(api3_match_check))
print("모든 운동 매칭:", api3_match_check["all_matched"].sum())
print("일부/미매칭:", (~api3_match_check["all_matched"]).sum())
```

```python
# 모든 구성 운동이 매칭된 API 3 영상 63개의 체력요인을 6개 가중치로 변환

api3_matched_file_nm = set(
    api3_match_check.loc[
        api3_match_check["all_matched"],
        "file_nm"
    ]
)

api3_factor_source = api3_current[
    api3_current["file_nm"].isin(api3_matched_file_nm)
][
    ["file_nm", "trng_nm", "matched_factor"]
].drop_duplicates()


def calculate_api3_video_weights(group):
    weights = {col: 0.0 for col in factor_cols}

    # 영상 안의 고유 운동별 체력요인을 한 번씩 반영
    for factor in group["matched_factor"]:
        if factor not in factor_mapping:
            continue

        for col, value in factor_mapping[factor].items():
            weights[col] += value

    total = sum(weights.values())

    if total > 0:
        weights = {
            col: value / total
            for col, value in weights.items()
        }

    return pd.Series(weights)


api3_factor_weights = (
    api3_factor_source
    .groupby("file_nm")
    .apply(calculate_api3_video_weights)
    .reset_index()
)

print("API 3 가중치 계산 영상:", len(api3_factor_weights))

print(
    "가중치 합이 1인 영상:",
    api3_factor_weights[factor_cols]
    .sum(axis=1)
    .round(10)
    .eq(1.0)
    .sum()
)

display(api3_factor_weights.head(10))
```

```python
# API 3에서 체력요인 매칭은 완료됐지만 6개 가중치 합이 1이 되지 않은 영상의 원인을 확인

api3_factor_weights["weight_sum"] = (
    api3_factor_weights[factor_cols]
    .sum(axis=1)
)

api3_invalid_weights = api3_factor_weights[
    ~api3_factor_weights["weight_sum"]
    .round(10)
    .eq(1.0)
].copy()

api3_invalid_detail = (
    api3_factor_source[
        api3_factor_source["file_nm"].isin(
            api3_invalid_weights["file_nm"]
        )
    ]
    .merge(
        api3_video[
            ["file_nm", "vdo_ttl_nm"]
        ],
        on="file_nm",
        how="left"
    )
)

print("가중치 합이 1이 아닌 영상:", len(api3_invalid_weights))

display(
    api3_invalid_weights[
        ["file_nm", "weight_sum"] + factor_cols
    ]
)

display(
    api3_invalid_detail[
        [
            "file_nm",
            "vdo_ttl_nm",
            "trng_nm",
            "matched_factor"
        ]
    ]
    .drop_duplicates()
    .sort_values(["file_nm", "trng_nm"])
)
```

```python
# 6개 체력요인 가중치 합이 1인 API 3 영상만 최종 체력요인 계산 결과로 확정

api3_factor_weights = api3_factor_weights[
    api3_factor_weights["weight_sum"]
    .round(10)
    .eq(1.0)
].copy()

api3_factor_weights = api3_factor_weights.drop(
    columns=["weight_sum"]
)

print("API 1 체력요인 확정:", 265)
print("API 3 체력요인 확정:", len(api3_factor_weights))
print("현재 총 확정 영상:", 265 + len(api3_factor_weights))
```

```python
# 체력요인이 확정되지 않은 나머지 70개 영상을 API별로 추출하여 확인

confirmed_file_nm = set(
    api1_factor_result.loc[
        api1_factor_result[factor_cols].notna().all(axis=1),
        "file_nm"
    ]
) | set(
    api3_factor_weights["file_nm"]
)

unresolved_videos = workout_videos[
    ~workout_videos["file_nm"].isin(confirmed_file_nm)
].copy()

print("전체 후보 영상:", len(workout_videos))
print("체력요인 확정:", len(confirmed_file_nm))
print("미확정 영상:", len(unresolved_videos))

print("\nAPI별 미확정 영상:")
print(
    unresolved_videos["source_api"]
    .value_counts()
)

display(
    unresolved_videos[
        [
            "source_api",
            "file_nm",
            "vdo_ttl_nm",
            "trng_nm",
            "ftns_fctr_nm"
        ]
    ].sort_values(
        ["source_api", "file_nm"]
    )
)
```

```python
# 미확정 API 3 영상 52개를 평형성·일부 매칭·전체 미매칭 유형으로 분류

api3_unresolved_file_nm = set(
    unresolved_videos.loc[
        unresolved_videos["source_api"] == "API 3",
        "file_nm"
    ]
)

api3_unresolved_detail = api3_current[
    api3_current["file_nm"].isin(api3_unresolved_file_nm)
].copy()

api3_unresolved_detail["six_factor_mappable"] = (
    api3_unresolved_detail["matched_factor"]
    .isin(factor_mapping.keys())
)

api3_unresolved_summary = (
    api3_unresolved_detail
    .groupby("file_nm")
    .agg(
        exercise_count=("trng_nm", "nunique"),
        matched_count=("matched_factor", lambda x: x.notna().sum()),
        six_factor_count=("six_factor_mappable", "sum"),
        matched_factors=(
            "matched_factor",
            lambda x: "|".join(
                sorted(set(x.dropna().astype(str)))
            )
        )
    )
    .reset_index()
)

def classify_unresolved(row):
    if (
        row["matched_count"] > 0
        and row["six_factor_count"] == 0
    ):
        return "6요인 외 체력요인"

    if row["six_factor_count"] == 0:
        return "전체 미매칭"

    if row["six_factor_count"] < row["exercise_count"]:
        return "일부 매칭"

    return "기타"


api3_unresolved_summary["unresolved_type"] = (
    api3_unresolved_summary.apply(
        classify_unresolved,
        axis=1
    )
)

print(
    api3_unresolved_summary["unresolved_type"]
    .value_counts()
)

display(
    api3_unresolved_summary[
        [
            "file_nm",
            "exercise_count",
            "matched_count",
            "six_factor_count",
            "matched_factors",
            "unresolved_type"
        ]
    ]
)
```

```python
# API 1과 API 3에서 계산이 완료된 영상별 6개 체력요인 가중치를 하나로 통합

api1_final_weights = api1_factor_result[
    api1_factor_result[factor_cols].notna().all(axis=1)
][
    ["file_nm"] + factor_cols
].copy()

api3_final_weights = api3_factor_weights[
    ["file_nm"] + factor_cols
].copy()

video_factor_weights = pd.concat(
    [api1_final_weights, api3_final_weights],
    ignore_index=True
)

print("체력요인 데이터:", len(video_factor_weights))
print("고유 영상:", video_factor_weights["file_nm"].nunique())
```

```python
# 최종 체력요인 데이터를 결합하고 실제 영상 URL을 포함한 12개 컬럼으로 정리

workout_videos = (
    workout_videos
    .drop(columns=factor_cols, errors="ignore")
    .merge(
        video_factor_weights,
        on="file_nm",
        how="inner"
    )
)

# 기존 기본 URL은 제거하고 완성된 영상 URL만 file_url로 사용
workout_videos = workout_videos.drop(
    columns=["file_url"],
    errors="ignore"
)

workout_videos = workout_videos.rename(
    columns={
        "vdo_ttl_nm": "title",
        "video_url": "file_url",
        "vdo_len": "video_length",
        "aggrp_nm": "age_group",
        "tool_nm": "equipment"
    }
)

workout_videos = workout_videos[
    [
        "file_nm",
        "title",
        "file_url",
        "video_length",
        "age_group",
        "strength",
        "muscularEndurance",
        "cardiovascularEndurance",
        "flexibility",
        "agility",
        "power",
        "equipment"
    ]
].copy()

print("영상 수:", len(workout_videos))
print("컬럼 수:", len(workout_videos.columns))
print("중복 컬럼:", workout_videos.columns.duplicated().sum())
```

```python
# 최종 운동 영상 CSV의 URL을 이용해 모든 영상을 로컬 폴더에 다운로드
import pandas as pd
import requests
from pathlib import Path

csv_path = Path("data/processed/workout_videos_v2.csv")
download_dir = Path("data/videos")

download_dir.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(csv_path)

for i, row in df.iterrows():
    file_nm = row["file_nm"]
    url = row["file_url"]
    save_path = download_dir / file_nm

    if save_path.exists():
        print(f"[{i+1}/{len(df)}] 이미 존재: {file_nm}")
        continue

    try:
        with requests.get(url, stream=True, timeout=60) as response:
            response.raise_for_status()

            with open(save_path, "wb") as f:
                for chunk in response.iter_content(chunk_size=1024 * 1024):
                    if chunk:
                        f.write(chunk)

        print(f"[{i+1}/{len(df)}] 완료: {file_nm}")

    except Exception as e:
        print(f"[{i+1}/{len(df)}] 실패: {file_nm} / {e}")
```
